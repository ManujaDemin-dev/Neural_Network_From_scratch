```python
import cv2
import os
import time 
import uuid
```

```python
IMAGES_PATH = 'Tensorflow/workspace/images/collectedimages'
```

```python
number_imgs = 15
```

```python
labels = ['A' , 'hello' , 'B']
```

```python
# Create folders
for label in labels:
    os.makedirs(os.path.join(IMAGES_PATH, label), exist_ok=True)

cap = cv2.VideoCapture(6)

for label in labels:

    print(f"\nCollecting images for {label}")
    print("Press SPACE to capture")
    print("Press N to go to next label")

    img_count = 0

    while img_count < number_imgs:

        ret, frame = cap.read()

        cv2.putText(frame, f'Label: {label}', (10,30),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

        cv2.putText(frame, f'Images: {img_count}/{number_imgs}', (10,70),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

        cv2.imshow('Image Collection', frame)

        key = cv2.waitKey(1) & 0xFF

        # SPACE = capture
        if key == 32:

            imgname = os.path.join(
                IMAGES_PATH,
                label,
                f'{label}.{uuid.uuid1()}.jpg'
            )

            cv2.imwrite(imgname, frame)
            img_count += 1

            print(f"Saved {imgname}")

        # N = skip label
        if key == ord('n'):
            break

        # Q = quit program
        if key == ord('q'):
            cap.release()
            cv2.destroyAllWindows()
            exit()

cap.release()
cv2.destroyAllWindows()
```
